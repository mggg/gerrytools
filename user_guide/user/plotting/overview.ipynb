The plotting module provides builder objects for statistical figures and geographic maps.
A builder keeps the data and display choices together: construct it, add datasets or map
layers, and then render or save the result. The figure can be revised without repeating the
original plotting call.


## Statistical plot builders

Choose a builder from the structure of the result, not from the number of input rows.

| Builder | Use it to show | Typical input |
| --- | --- | --- |
| `Histogram` | Frequency of one or more outcomes | One value per plan |
| `BarPlot` | Values across a small set of categories | One value per category |
| `BoxPlot` | Quantiles across categories or district ranks | Many values per category |
| `ViolinPlot` | Distribution shape across categories or ranks | Many values per category |
| `ScatterPlot` | Relationship between two quantities | Paired x and y values |
| `SeaLevelPlot` | Many plans across ordered categories | One value per plan and category |
| `SeatsVotesPlot` | Response of seats to uniform vote swing | District votes and totals |
| `PaintballPlot` | Outcomes in vote-share and seat-share space | Paired statewide shares |


```python
from gerrytools.plotting import Histogram

plot = Histogram(legend=True)
plot.add_dataset([1] + [2] * 4 + [3] * 6 + [4] * 4 + [5])
plot.show()
```

Each format guide begins with the constructor and the primary `add_*` method, then shows the
comparison layers that are specific to that plot. Repeated `add_*` calls usually add another
dataset to the same axes. The [statistical plot index](statistical/index.md) links to each
format and to a combined workflow.

## Geographic plot builders

`GeoPlot` draws plans, choropleths, outlines, highlights, markers, and labels from a
GeoDataFrame. Each `add_*_layer()` call contributes one layer, and `zorder` determines how
overlapping layers stack. `DotDensityPlot` uses the same model for
randomly placed population symbols. The [geographic plot guide](geographic/geo.ipynb) shows
the main layers independently before the workflow guide combines them.


## Render and save

`show()` renders the current builder in a notebook or interactive session. `save()` renders
the same state and writes it to a file. Accessing `.ax` also triggers a render and returns the
Matplotlib axes for formatting or composition that GerryTools does not expose directly.

The plot builders are "lazy." This means adding data or changing an option marks the figure for
rebuilding, and the next access to `.ax`, `show()`, or `save()`, or a call to `bind_to_ax()`, 
applies the current state. Add ordinary Matplotlib artists after the final GerryTools change when 
those artists must remain on top.


```python
axes = plot.ax
type(axes).__name__
```

## Add a builder to an existing figure

In an effort to support common Matplotlib workflows, the statistical plot builders accept an `ax=` 
parameter. This allows for experienced Matplotlib users to enjoy the ergonomics of the provided 
builder classes while maintaining the power of the base library. Because of the rendering schema
for Geographic plot builders, however, they require that a user call `bind_to_ax()` to render the
plot onto existing axes. For a complete multi-panel example, see 
[Matplotlib composition](composition.ipynb).


```python
import matplotlib.pyplot as plt

figure, axes = plt.subplots()
embedded = Histogram(ax=axes)
embedded.add_dataset([1] + [2] * 6 + [3] * 15 + [4] * 20 + [5] * 15 + [6] * 6 + [7])
embedded.ax;
```
