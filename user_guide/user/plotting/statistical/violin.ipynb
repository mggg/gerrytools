`ViolinPlot` shows a smoothed distribution for each category. Compared with a box plot, it
makes skew, shoulders, and multiple concentrations visible instead of reducing each
distribution to quantiles.

```python
import json
from pathlib import Path

import numpy as np

from gerrytools.plotting import ViolinPlot

data_dir = Path("data")
records_a = json.loads((data_dir / "ga_congressional_ensemble_a.json").read_text())
records_b = json.loads((data_dir / "ga_congressional_ensemble_b.json").read_text())
ranked = np.sort(np.asarray([record["BVAP"] for record in records_a]), axis=1)
rank_labels = [str(rank) for rank in range(1, ranked.shape[1] + 1)]
distributions = {label: ranked[:, index] for index, label in enumerate(rank_labels)}
```

```python
plot = ViolinPlot()
plot.add_dataset(distributions, facecolor="denim")
plot.show()
```

Each violin is estimated from 1,000 plans. Its width represents density, not the number of
observations. As with the box plot, each x-position is a district rank rather than a fixed
district identity.

## Add a comparison plan

`add_pointset()` uses the same category mapping as the distribution.

```python
starting_plan = dict(zip(rank_labels, ranked[0], strict=True))
plot.add_pointset(starting_plan, facecolor="applegreen")
plot.show()
```

## Multiple datasets

Repeated `add_dataset()` calls place several violin families side by side. This compares the
same ranked BVAP measure across the two ensemble runs.

```python
ranked_b = np.sort(np.asarray([record["BVAP"] for record in records_b]), axis=1)
distributions_b = {label: ranked_b[:, index] for index, label in enumerate(rank_labels)}

comparison = ViolinPlot()
comparison.add_dataset(distributions, facecolor="denim")
comparison.add_dataset(distributions_b, facecolor="cherryblossompink")
comparison.show()
```

The input may be a category mapping, nested lists, or a DataFrame. Shared labels, legends,
limits, reference lines, and output controls live in
[shared statistical controls](options.ipynb).
