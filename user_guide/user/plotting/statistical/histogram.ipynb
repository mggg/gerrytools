The `Histogram` class builds a basic histogram. In GerryTools, we have elected to treat histograms
and barplots as separate entities.

```python
import numpy as np

from gerrytools.plotting import Histogram

rng = np.random.default_rng(42)

example_data = rng.binomial(10, 0.5, size=1000)
```

Creating a histogram follows the basic syntax workflow: instantiate the class, add the data, show
the plot.

```python
plot = Histogram()
plot.add_dataset(example_data)
plot.show()
```

Matplotlib controls the normal binning, but since these are integer outcomes, so we may use 
`set_bin_widths(1)` to get a better plot:

```python
plot = Histogram()
plot.add_dataset(example_data)
plot.set_bin_widths(1)
plot.show()
```

And we may shift the bars leftwards by half thier width for centering if so desired:

```python
plot = Histogram()
plot.add_dataset(example_data)
plot.set_bin_widths(1)
plot.center_bars()
plot.set_xticks(range(1, 10))
plot.show()
```

## Add a comparison value

Reference lines and point markers distinguish a plan from the ensemble distribution without
turning it into a second histogram.

```python
plot.add_vertical_lines(4, linecolor="cherryblossompink", linewidth=2)
plot.show()
```

## Compare distributions

Repeated `add_dataset()` calls can overlay, group, stack, or outline distributions according
to `histtype`. Use `as_density()` when distributions with different sample sizes should be
compared by proportion instead of raw count.

```python
example_data2 = rng.binomial(10, 0.4, size=1000)

comparison = Histogram()
comparison.add_dataset(example_data, name="Sample 1")
comparison.add_dataset(example_data2, facecolor="citizen_blue", facealpha=0.65, name="Sample 2")
comparison.set_bin_widths(1)
comparison.center_bars()
comparison.set_xticks(range(0, 11))
comparison.set_xlim(-1.5, 11.5)
comparison.show()
```

`add_dataset()` accepts an iterable, NumPy array, Series, or one DataFrame column. `weights=`
plots an aggregated value-count table without expanding it into repeated observations. Shared
axes, labels, legends, and output controls are covered in
[shared statistical controls](options.ipynb).
