`BarPlot` compares values across a small number of categories. Here the categories are two
statewide elections, and the values are district-level Democratic seat counts.

```python
from gerrytools.plotting import BarPlot

democratic_seats = {"President": 12, "U.S. Senate": 8}
```

```python
plot = BarPlot()
plot.add_dataset(democratic_seats, facecolor="denim")
plot.show()
```

A mapping carries its category labels with the values. `add_dataset()` also accepts a list,
Series, or DataFrame; lists use `labels=` and DataFrames use `column=`.

## Grouped and stacked data

Repeated `add_dataset()` calls draw grouped bars by default. 

```python
republican_seats = {
    "President": 6,
    "U.S. Senate": 10,
}

dem_rep_barplot = BarPlot()
dem_rep_barplot.add_dataset(democratic_seats, facecolor="denim")
dem_rep_barplot.add_dataset(republican_seats, facecolor="alizarin")
dem_rep_barplot.show()
```

Construct the builder with `stacked=True` or set `plot.stacked=True` to stack the plots. 

```python
dem_rep_barplot.stacked = True
dem_rep_barplot.show()
```
