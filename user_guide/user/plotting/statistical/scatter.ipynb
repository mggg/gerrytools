`ScatterPlot` shows the relationship between two quantities. Here, each point represents one
Georgia congressional plan. Its horizontal position is the plan's mean Democratic
disproportionality across thirteen statewide elections, and its vertical position is the
variance of those scores. This follows the ensemble comparison in
[Redistricting for Proportionality](https://doi.org/10.1515/for-2022-2064).

```python
from pathlib import Path

import pandas as pd

from gerrytools.plotting import ScatterPlot

data_dir = Path("data")
scores = pd.read_parquet(data_dir / "ga_congressional_disproportionality_100000.parquet").set_index(
    "step"
)
elections = tuple(scores.columns)
mean_disprop = scores.mean(axis=1)
variance = scores.var(axis=1, ddof=0)
```

```python
plot = ScatterPlot(legend=False)
plot.add_series(
    mean_disprop,
    variance,
    markerfacecolor="default_grey",
    markersize=3,
)
plot.add_vertical_lines(0)
plot.set_xlim(-0.3, 0.3)
plot.set_ylim(0, 0.02)
plot.show()
```

Each election score is Democratic seat share minus statewide two-party Democratic vote share.
A negative mean therefore indicates that, on average, a plan gives Democrats a smaller share
of seats than votes. The variance measures how much that result changes across the thirteen
elections: plans near the bottom behave more consistently, while plans higher in the plot are
more sensitive to the election used. Some points coincide because seat shares, and therefore
disproportionality scores, are discrete. `xy_pairs=` also accepts a sequence of `(x, y)` tuples.

## Add one comparison plan

`add_point()` places a named observation over the existing series. The shared controls guide
shows how names become legend entries; the plot below turns the legend on.

```python
plot = ScatterPlot(legend=True)
plot.add_series(
    mean_disprop,
    variance,
    "Ensemble Values",
    markerfacecolor="default_grey",
    markerfacealpha=0.7,
    markersize=3,
)
plot.add_point(
    -0.052,
    0.0038,
    "Example Plan",
    markerfacecolor="amber",
    markeredgecolor="black",
    markersize=9,
)
plot.add_vertical_lines(0)
plot.set_xlim(-0.3, 0.3)
plot.set_ylim(0, 0.02)
plot.show()
```

Reference lines, bands, limits, labels, and legends are covered in
[shared statistical controls](options.ipynb).
