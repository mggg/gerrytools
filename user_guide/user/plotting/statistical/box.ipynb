`BoxPlot` shows a distribution for each category. For a marginal ensemble plot, sort every
plan's districts from lowest to highest, then collect the values at each rank. A rank is a
position within each plan, not one district followed through the ensemble.

```python
import json
from pathlib import Path

import numpy as np

from gerrytools.plotting import BoxPlot

records_a = json.loads(Path("data/ga_congressional_ensemble_a.json").read_text())
records_b = json.loads(Path("data/ga_congressional_ensemble_b.json").read_text())
ranked_bvap = np.sort(np.asarray([record["BVAP"] for record in records_a]), axis=1)
rank_labels = [str(rank) for rank in range(1, ranked_bvap.shape[1] + 1)]
bvap_by_rank = {label: ranked_bvap[:, index] for index, label in enumerate(rank_labels)}
```

The mapping passed to `add_dataset()` contains 1,000 observations for each of Georgia's 14
congressional district ranks.

```python
plot = BoxPlot()
plot.add_dataset(bvap_by_rank, facecolor="default_grey")
plot.show()
```

Read vertically to see the ensemble spread at one rank and horizontally to see how quickly
the district values rise within plans. `add_pointset()` overlays a comparison plan using the
same category labels.

```python
starting_plan = dict(zip(rank_labels, ranked_bvap[0], strict=True))
plot.add_pointset(starting_plan, facecolor="cherryblossompink")
plot.show()
```

## Multiple datasets

A box plot can hold several datasets at once. Repeated `add_dataset()` calls place their boxes
side by side at each rank. Here the same BVAP ranks are compared across two ensemble runs.

```python
ranked_bvap_b = np.sort(np.asarray([record["BVAP"] for record in records_b]), axis=1)
bvap_b_by_rank = {label: ranked_bvap_b[:, index] for index, label in enumerate(rank_labels)}

comparison = BoxPlot()
comparison.add_dataset(bvap_by_rank, facecolor="default_grey")
comparison.add_dataset(bvap_b_by_rank, facecolor="denim")
comparison.show()
```

`add_dataset()` also accepts nested sequences, a DataFrame, or a NumPy array. Use
`add_stats_dataset()` with `BoxPlotStats` when only precomputed quartiles and whiskers are
available. Percentile rules, fliers, labels, legends, and axes are covered in
[shared statistical controls](options.ipynb).
