GerryTools plots expose their Matplotlib axes through `.ax`. Statistical builders accept an
existing axes with `ax=`. Geographic builders are configured first and then moved with
`bind_to_ax()`. The builder owns its plotted artists, while the caller owns the surrounding
figure and layout.

```python
import random

import matplotlib.pyplot as plt

from gerrytools.plotting import Histogram

rng = random.Random(42)

figure, axes = plt.subplots(1, 2, figsize=(9, 3))

left = Histogram(ax=axes[0])
left.add_dataset([rng.randint(1, 5) for _ in range(10)], edgecolor="black")
left.set_bin_widths(1)

right = Histogram(ax=axes[1])
right.add_dataset([rng.randint(1, 5) for _ in range(10)], edgecolor="black")
right.set_bin_widths(1)

left.ax
right.ax;
```

Accessing `left.ax` and `right.ax` renders each builder into its assigned panel.

The caller controls the supplied figure's size, layout, display, and closing. Configure the
builder before adding ordinary Matplotlib artists because later `.ax`, `show()`, or `save()`
calls rebuild the GerryTools-managed artists. Legends default to an outside-right anchor, so
multi-panel layouts should either move each legend inside its axes or reserve space for it.

```python
plot = Histogram(legend=True)
plot.add_dataset([1, 1, 2, 3, 5, 8], edgecolor="black")
plot.set_bin_widths(1)
plot.add_vertical_lines(
    3.5,
    linecolor="cherryblossompink",
    linestyle="--",
    linewidth=2,
    name="GerryTools Line",
)
plot.ax.figure
```

## Moving an existing plot

`bind_to_ax()` moves a configured builder to another axes, immediately renders its current
state, and preserves its data and settings. Pass `None` to return it to a GerryTools-owned
figure.

```python
new_figure, new_axes = plt.subplots(figsize=(5, 3))
plot.bind_to_ax(new_axes)
```
