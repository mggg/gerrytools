`draw_graph()` draws a NetworkX graph using x- and y-coordinate node attributes. This example
builds the adjacency graph of Georgia's 14 congressional districts and places each node near
the district it represents.

```python
from pathlib import Path

import geopandas as gpd
import matplotlib.pyplot as plt
import networkx as nx
from gerrychain import Graph

from gerrytools.colors import districtr
from gerrytools.plotting import draw_graph

precincts = gpd.read_file(Path("data/ga_2016_precincts.gpkg"))
districts = precincts[["CD", "geometry"]]
points = districts.geometry.representative_point()

graph = Graph.from_geodataframe(districts).to_networkx_graph()
nx.set_node_attributes(graph, points.x.to_dict(), "X")
nx.set_node_attributes(graph, points.y.to_dict(), "Y")
```

The graph records adjacency; the coordinate attributes only determine where the nodes are
drawn. Edges are straight schematic connections rather than shared boundary segments.

```python
n_districts = districts["CD"].nunique()
node_colors_list = districtr(n_districts)
node_colors = [node_colors_list[int(d) - 1] for d in districts["CD"]]

fig, ax = plt.subplots(dpi=300)
draw_graph(
    graph, x="X", y="Y", node_color=node_colors, node_size=5, edge_color="#808080", width=0.5, ax=ax
)
plt.show()
```

The default coordinate names are `INTPTLON20` and `INTPTLAT20`, which are common in Census
geography files. Extra keyword arguments are passed to NetworkX, including `node_color`,
`edge_color`, `node_size`, `width`, and `with_labels`.

## Disconnected graphs

`draw_graph_components()` returns one `(figure, axes)` pair for each connected component. This
is useful for graph inputs containing islands or several disconnected study regions.
