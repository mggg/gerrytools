`DotDensityPlot` places a fixed number of randomly located dots inside each source polygon.
Each dot represents `people_per_dot` people; it does not represent an exact residential
location.

```python
from pathlib import Path

import geopandas as gpd
import pandas as pd

from gerrytools.plotting import DotDensityPlot

precincts = gpd.read_file(Path("data/ga_2016_precincts.gpkg"))
columns = ["NH_WHITE", "NH_BLACK", "HISP"]
precincts[columns] = precincts[columns].apply(pd.to_numeric)
```

Add one density layer for each count column. `outline_column` draws the district boundaries.

```python
plot = DotDensityPlot(
    precincts,
    outline_column="CD",
    people_per_dot=1_000,
    rng_seed=2026,
    silent=True,
)
plot.add_density_layer("NH_WHITE", "lightsteelblue", n_jobs=1)
plot.add_density_layer("NH_BLACK", "alizarin", n_jobs=1)
plot.add_density_layer("HISP", "amber", n_jobs=1)
plot.show()
```

## Reusing generated dots

Generated points are cached for the lifetime of the plot, so changing marker, outline, or
focus settings does not regenerate them. Pass `refresh_cache=True` when the geometry or count
values have changed. `n_jobs` and `n_chunks` control parallel dot generation for larger maps.

Dot counts are rounded separately for each source polygon. Keep `people_per_dot` fixed when
comparing maps.

```python
metro = precincts["CTYNAME"].isin(["Clayton", "Cobb", "DeKalb", "Fulton", "Gwinnett"])
plot.focus_axes(geometry_mask=metro, pad=0.03)
plot.show()
```
